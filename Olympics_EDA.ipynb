```python
# Load dataset
df = pd.read_csv("Summer-Olympic-medals-1976-to-2008.csv", encoding="latin1")


```

```python
# Display basic information
print(df.info())

# Show first few rows
print(df.head())

# Check for missing values
print(df.isnull().sum())

# Check for duplicate values
print(df.duplicated().sum())
```

```python
# Drop duplicate rows
df = df.drop_duplicates()

# Fill missing values if necessary
df.fillna(method='ffill', inplace=True)  # Example: Forward fill missing values

# Convert data types (if required)
# df['Year'] = df['Year'].astype(int)  # Example if Year is not in integer format

print(df.info())  # Check after cleaning
```

```python
# Top 10 countries with the most medals
top_countries = df['Country'].value_counts().head(10)
plt.figure(figsize=(10,5))
sns.barplot(x=top_countries.index, y=top_countries.values, palette='viridis')
plt.title("Top 10 Countries with Most Medals")
plt.xlabel("Country")
plt.ylabel("Medal Count")
plt.xticks(rotation=45)
plt.show()
```

```python
# Year-wise distribution of medals
plt.figure(figsize=(10,5))
sns.countplot(x='Year', data=df, palette='coolwarm')
plt.title("Medal Count by Year")
plt.xlabel("Year")
plt.ylabel("Count")
plt.show()
```

```python
# Top 10 most successful athletes
top_athletes = df['Athlete'].value_counts().head(10)

plt.figure(figsize=(10,5))
sns.barplot(x=top_athletes.index, y=top_athletes.values, palette='magma')
plt.xticks(rotation=45)
plt.title("Top 10 Athletes with Most Medals")
plt.xlabel("Athlete")
plt.ylabel("Total Medals")
plt.show()
```

```python
top_sports = df['Sport'].value_counts().head(10)
plt.figure(figsize=(10,5))
sns.barplot(x=top_sports.index, y=top_sports.values, palette='Reds')
plt.xticks(rotation=45)
plt.title("Top 10 Sports with Most Medals")
plt.xlabel("Sport")
plt.ylabel("Total Medals")
plt.show()
```

```python
# Gender distribution in Olympic medals
plt.figure(figsize=(6,4))
sns.countplot(x='Gender', data=df, palette='pastel')
plt.title("Medal Distribution by Gender")
plt.xlabel("Gender")
plt.ylabel("Count")
plt.show()
```

```python
# Grouping data by Country and Year to count medals
country_year_medals = df.groupby(["Year", "Country"])["Medal"].count().reset_index()

# Pivot for visualization
country_pivot = country_year_medals.pivot(index="Year", columns="Country", values="Medal").fillna(0)

# Plot
plt.figure(figsize=(12, 6))
sns.heatmap(country_pivot, cmap="coolwarm", linewidths=0.5)
plt.title("Country-wise Medal Count Over the Years (Heatmap)")
plt.xlabel("Country")
plt.ylabel("Year")
plt.show()
```

```python
# Total medals per country
total_medals_country = df["Country"].value_counts()

# Selecting top 10
top_10_countries = total_medals_country.head(10)

# Pie Chart
plt.figure(figsize=(10, 6))
plt.pie(top_10_countries, labels=top_10_countries.index, autopct="%1.1f%%", colors=sns.color_palette("Paired", 10))
plt.title("Medal Share of Top 10 Countries (1976-2008)")
plt.show()
```

```python
# Count of medals by country and sport
dominant_sport_per_country = df.groupby(["Country", "Sport"])["Medal"].count().reset_index()

# Find the top sport per country
dominant_sport_per_country = dominant_sport_per_country.sort_values("Medal", ascending=False).groupby("Country").head(1)

print(dominant_sport_per_country)
```

```python
# Count total athletes per event
athlete_counts = df.groupby(["Year", "Event"])["Athlete"].count().reset_index(name="Total_Finalists")

# Count medal-winning athletes per event
medal_winners = df.groupby(["Year", "Event"])["Medal"].count().reset_index(name="Medalists")

# Merge both datasets
conversion_df = pd.merge(athlete_counts, medal_winners, on=["Year", "Event"], how="left")

# Calculate Conversion Rate
conversion_df["Conversion Rate"] = (conversion_df["Medalists"] / conversion_df["Total_Finalists"]) * 100

# Display results
print(conversion_df.head(10))
```

```python
# Count of Olympics participated per athlete
athlete_appearances = df.groupby("Athlete")["Year"].nunique().reset_index(name="Olympics_Participated")

# Athletes with most appearances
top_athletes = athlete_appearances.sort_values(by="Olympics_Participated", ascending=False).head(10)

print(top_athletes)
```

```python
# Count athletes by gender per year
gender_distribution = df.groupby(["Year", "Gender"])["Athlete"].nunique().reset_index()

# Pivot for visualization
gender_pivot = gender_distribution.pivot(index="Year", columns="Gender", values="Athlete")

# Plot
plt.figure(figsize=(12, 6))
gender_pivot.plot(kind="line", marker="o", colormap="coolwarm")
plt.xlabel("Year")
plt.ylabel("Number of Athletes")
plt.title("Gender Representation Over Time (1976-2008)")
plt.legend(title="Gender")
plt.show()
```
